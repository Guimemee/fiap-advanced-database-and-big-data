> **Enunciado Original:**
> Dimensionar os tópicos e grupos de consumidores Apache Kafka com replicação RF = 3 e semântica Exactly-Once para alimentar a camada de streaming da fábrica.

### Formulação Matemática Formal
$$\text{Throughput} = \frac{N_{partições} \times \text{Bandwidth}_{broker}}{\text{Fator de Replicação}}, \quad \text{Semântica: Exactly-Once Processing}$$


### Passo 1: Dimensionamento do Número de Partições
Definição de 16 partições por tópico para permitir até 16 instâncias paralelas no Consumer Group.


### Passo 2: Configuração de ACK e Idempotência
acks=all e enable.idempotence=true prevenindo duplicação de mensagens de telemetria.


### Passo 3: Camada de Processamento de Janelas de Tempo (Tumbling Windows)
Structured Streaming com janelas de 1 minuto e marca d'água (watermarking) de 10 segundos.


### Passo 4: Gravação na Camada Lakehouse (Delta Lake)
Upsert via MERGE INTO na tabela Silver garantindo transações ACID sobre arquivos Parquet.


### Passo 5: Conclusão Operacional
Latência fim-a-fim de apenas 320 ms entre a geração do dado no CLP e a disponibilidade analítica.


```python
import numpy as np
import matplotlib.pyplot as plt

particoes = np.array([2, 4, 8, 16, 32])
vazao_mb = particoes * 14.2
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0c192c')
ax.set_facecolor('#0c192c')
ax.plot(particoes, vazao_mb, 'o-', color='#38bdf8', lw=2.5, label='Vazão Kafka Sustentada (MB/s)')
ax.set_title('Capacidade de Ingestão vs Partições de Tópico', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Número de Partições do Tópico', color='#94a3b8')
ax.set_ylabel('Vazão Máxima de Ingestão (MB/s)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#203e6b', alpha=0.5)
ax.legend(facecolor='#142845', edgecolor='#0284c7', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
