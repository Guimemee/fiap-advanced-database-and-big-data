> **Enunciado Original:**
> Implementar pipeline completo Medallion Architecture para transformar dados brutos não estruturados (Bronze), aplicar saneamento e tipagem (Silver) e agregar métricas de KPIs (Gold).

### Formulação Matemática Formal
$$\text{Bronze (Raw JSON)} \xrightarrow{\text{Data Cleaning}} \text{Silver (Clean Parquet)} \xrightarrow{\text{Business Agg}} \text{Gold (Curated KPIs)}$$


### Passo 1: Ingestão Raw na Camada Bronze
Armazenamento de payloads JSON brutos imutáveis com carimbo de data/hora de ingestão.


### Passo 2: Higienização e Validação na Camada Silver
Descarte de registros nulos, conversão estrita de tipos e validação de limites físicos.


### Passo 3: Otimização de Armazenamento com Z-ORDER
Execução de OPTIMIZE silver_telemetria ZORDER BY (id_equipamento, data) para compactação de arquivos pequenos.


### Passo 4: Agregações de Negócio na Camada Gold
Tabelas sumarizadas de OEE, disponibilidade e consumo energético para alimentação de Power BI.


### Passo 5: Conclusão do Desafio Sanofi
Lakehouse corporativo unificado suportando tanto BI analítico quanto modelos de IA preditiva.


```python
import numpy as np
import matplotlib.pyplot as plt

camadas = ['Bronze (Bruto)', 'Silver (Limpo)', 'Gold (Curado)']
tamanho_gb = [140, 48, 4.2]
cores = ['#94a3b8', '#0284c7', '#10b981']
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0c192c')
ax.set_facecolor('#0c192c')
bars = ax.bar(camadas, tamanho_gb, color=cores, width=0.45)
ax.set_title('Redução de Volume e Compactação na Arquitetura Medallion', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Volume de Armazenamento (GB)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#203e6b', alpha=0.5)
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 2, f'{yval} GB', ha='center', va='bottom', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
