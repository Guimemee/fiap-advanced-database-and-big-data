> **Enunciado Original:**
> Modelar coleções JSON flexíveis e construir pipelines de agregação (Aggregation Framework: $match, $group, $unwind, $project) para análise de telemetria dos reatores Sanofi.

### Formulação Matemática Formal
$$\text{db.reatores.aggregate}([\{\$match: \dots\}, \{\$group: \dots\}, \{\$project: \dots\}])$$


### Passo 1: Estratégia de Incorporação vs Referenciamento
Documentos embutidos (Embedding) para leituras atômicas frequentes de sensores por lote.


### Passo 2: Criação de Índices Multikey e Índices TTL
Índices compostos {id_reator: 1, timestamp: -1} para suporte a consultas em janela deslizante.


### Passo 3: Estágios de Agregação Analítica
$match filtra anomalias térmicas; $group calcula média e desvio padrão em blocos de 5 minutos.


### Passo 4: Projeção de Dados Enriquecidos
$project reformata o esquema JSON para consumo imediato por dashboards operacionais.


### Passo 5: Conclusão de Throughput
Taxa de escrita sustentada de 42.000 documentos/s sem bloqueio transacional.


```python
import numpy as np
import matplotlib.pyplot as plt

estagios = ['$match (Filtro)', '$group (Agregação)', '$project (Saída)']
docs = [100000, 1200, 1200]
cores = ['#0284c7', '#2563eb', '#38bdf8']
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0c192c')
ax.set_facecolor('#0c192c')
bars = ax.bar(estagios, docs, color=cores, width=0.45)
ax.set_yscale('log')
ax.set_title('Funil de Agregação MongoDB (Documentos Processados)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_ylabel('Quantidade de Documentos (Log)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='y', color='#203e6b', alpha=0.5)
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
