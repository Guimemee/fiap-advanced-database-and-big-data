```python
import numpy as np
import matplotlib.pyplot as plt

n_linhas = np.array([1, 5, 10, 25, 50])
t_serial = n_linhas * 0.28
t_spark = 0.5 + n_linhas * 0.018
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#0c192c')
ax.set_facecolor('#0c192c')
ax.plot(n_linhas, t_serial, 'o--', color='#ef4444', lw=2, label='Consulta Tradicional RDBMS')
ax.plot(n_linhas, t_spark, 's-', color='#38bdf8', lw=2.5, label='SQL Otimizado Distribuído')
ax.set_title('Tempo de Processamento vs Volume (Milhões de Linhas)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Milhões de Registros', color='#94a3b8')
ax.set_ylabel('Tempo de Resposta (segundos)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#203e6b', alpha=0.5)
ax.legend(facecolor='#142845', edgecolor='#0284c7', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
